```python
import pandas as pd
from datetime import datetime, timedelta
import numpy as np
import warnings
warnings.filterwarnings('ignore')
```

```python
"""Load output data from Excel file with multiple sheets"""
# 2 files to ensemble
file_path_1 = "./Phase 1 Prediction Output_original_ezra.xlsx"
file_path_2 = "./Phase 1 Predictions Output_Fannisa.xlsx"
output_path = "./Phase 1 Ensemble Output.xlsx"
xl = pd.ExcelFile(file_path_1)
groups = xl.sheet_names

def ensemble(groups, file_path_1, file_path_2, output_path, *kwargs):
    # Create Excel writer
    writer = pd.ExcelWriter(output_path, engine='openpyxl')
    for sheet in groups:
        df1 = pd.read_excel(file_path_1, sheet_name=sheet)
        df2 = pd.read_excel(file_path_2, sheet_name=sheet)
        # # Skip the first row which contains staff type labels
        # df1 = df1.iloc[1:]
        df2 = df2.iloc[1:]
        # Convert the date column to datetime
        df1['Date'] = pd.to_datetime(df1.iloc[:, 0])
        df2['Date'] = pd.to_datetime(df2.iloc[:, 0])
        # Sort by date to ensure temporal order
        df1 = df1.sort_values('Date')
        df2 = df2.sort_values('Date')
        # Exclude 'Date' column and compute the mean of the numeric columns
        df_mean = (df1.drop(columns='Date') + df2.drop(columns='Date')) / 2
        
        # Add the Date column back
        df_mean['Date'] = df1['Date']
        # Create DataFrame and save to sheet
        df_mean.to_excel(writer, sheet_name=sheet, index=False)
        display(df_mean)
    writer.close()        

        
```

```python
ensemble(groups, file_path_1, file_path_2, output_path)
```
